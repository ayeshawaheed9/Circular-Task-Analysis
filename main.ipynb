```python
# EXPORT-CELL
# Saves the notebook as main.html next to it and opens it in the browser.
# How to use: Run All, press Ctrl+S to save (this keeps the outputs), then run THIS cell again.
# Cells tagged "hide" and the cell numbers are left out of the HTML.
import os, webbrowser
import nbformat
from nbconvert import HTMLExporter
from nbconvert.preprocessors import TagRemovePreprocessor

nb_path = os.path.abspath(globals().get("__vsc_ipynb_file__", "main.ipynb"))
nb = nbformat.read(nb_path, as_version=4)
nb.cells = [c for c in nb.cells if "# EXPORT-CELL" not in c.source]

if not any(c.cell_type == "code" and c.outputs for c in nb.cells):
    print("The saved notebook has no outputs yet. Run All, press Ctrl+S, then run this cell again.")
else:
    exporter = HTMLExporter(exclude_input_prompt=True, exclude_output_prompt=True)
    exporter.register_preprocessor(TagRemovePreprocessor(remove_cell_tags={"hide"}), enabled=True)
    body, _ = exporter.from_notebook_node(nb)
    html_path = os.path.splitext(nb_path)[0] + ".html"
    with open(html_path, "w", encoding="utf-8") as f:
        f.write(body)
    webbrowser.open("file://" + html_path)
    print("Saved", html_path)
```
